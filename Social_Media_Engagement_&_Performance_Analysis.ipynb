# **Social Media Engagement & Performance Analysis**

## **Objectives**

The main objective of the Social Media Engagement & Performance Analysis is to analyze social media data and understand the factors that influence audience engagement and content performance. The project examines engagement metrics such as likes, comments, shares, impressions, watch time, follower count, and engagement rate to identify meaningful patterns and performance differences across various segments.

## **Problem Statement**

The increasing volume of social media data makes it difficult to understand audience behavior and determine which factors contribute to better engagement. The major problems addressed in this project are:

- To analyze the overall engagement and performance of social media content.
- To understand variations in likes, comments, shares, impressions, watch time, and engagement rate.
- To identify performance differences among different content types and categories.
- To examine engagement patterns across countries and different age groups.
- To analyze the influence of sentiment and device type on engagement behavior.
- To identify relationships between numerical engagement metrics using correlation, pair plots, and other visualizations.
- To detect unusual or extreme engagement values that may affect the analysis.
- To identify important patterns that can support better content planning and social media performance improvement.

```python
# * Heatmap: correlation matrix

plt.figure(figsize=(10, 7))
sns.heatmap(corr_df,annot=True)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()
plt.show()
```

**Insight:** The correlation heatmap demonstrates that social-media engagement is not strongly explained by any single numerical variable. The strongest observed relationship is the **negative correlation** between **impression count** and **engagement rate** (r = -0.2322), while **likes** show only a very **weak positive relationship** (r = 0.0935). Comments, shares, watch time, follower count, and age have correlations **close to zero** with engagement rate. This indicates that reach, audience size, and individual engagement actions do not independently explain overall engagement efficiency. Therefore, social-media performance should be analyzed using a combination of content, audience, sentiment, timing, reach, and engagement characteristics rather than relying on one KPI.

```python
# * Swarm plot: engagement vs device

sample = df.sample(min(1000, len(df)), random_state=42)

plt.figure(figsize=(9, 5))
sns.swarmplot(data=sample, x="device_type", y="engagement_rate", size=3)
plt.title("Engagement Rate by Device Type")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

**Insight:** The swarm plot demonstrates that engagement rate varies substantially across individual observations within each device category. Mobile users have the highest average watch time, but watch time has almost **no** **linear correlation** with **engagement rate** in the dataset. Therefore, device type and watch time should be considered supporting features rather than standalone explanations of engagement.

**Mobile** shows a higher concentration of engagement rates compared with the other devices, indicating stronger audience engagement among posts accessed through that device.  But the difference between other devices are relatively small.

```python
# Step1 - Using IQR method

df1=df.copy()
```

```python
Q1=df1['engagement_rate'].quantile(0.25)
Q3=df1['engagement_rate'].quantile(0.75)
IQR=Q3-Q1
UB=IQR+1.5*Q3
LB=IQR-1.5*Q1

print(Q1,Q3,IQR,UB,LB,sep="\n")
```

```python
df1[(df1['engagement_rate']<LB) | (df1['engagement_rate']>UB)]
```

**Note:** Using the IQR method, 1,728 rows were identified as outliers. Removing all these rows may result in significant data loss.

```python
# Step2 - Cap Outliers (Winsorization) method

from scipy.stats.mstats import winsorize

df['engagementrate_winsor'] = winsorize(df['engagement_rate'], limits=[0.05, 0.05])
```

```python
df['engagementrate_winsor'].kurtosis()
```

```python
df['engagementrate_log_winsor'] = np.log1p(df['engagementrate_winsor'])
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))

sns.histplot(df['engagement_rate'], kde=True, ax=ax[0])
ax[0].set_title("Original")

sns.histplot(df['engagementrate_log_winsor'], kde=True, ax=ax[1])
ax[1].set_title("After Log winsor Transform")

plt.show()
```
